# Logistics ML — Linear Regression
This practical notebook loads the logistics CSV, preprocesses the numeric features, trains Linear Regression, evaluates it, visualizes predictions, and exports the model used by Flask.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import joblib
```

## 1. Load and inspect the dataset

```
df = pd.read_csv('../data/logistics.csv')
display(df.head())
print('Shape:', df.shape)
print('\nInfo:')
df.info()
print('\nMissing values:')
print(df.isnull().sum())
```

## 2. Basic preprocessing and feature/target split

```
feature_columns = [
    'distance_km','package_weight_kg','traffic_level','weather_score',
    'vehicle_age_years','driver_experience_years','warehouse_delay_hours',
    'loading_time_hours','priority_level'
]
target_column = 'delivery_time_hours'
df = df.dropna(subset=feature_columns + [target_column]).copy()
X = df[feature_columns]
y = df[target_column]
print('Feature order:', feature_columns)
print('Target:', target_column)
```

## 3. Train/test split and Linear Regression

```
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
model = LinearRegression()
model.fit(X_train, y_train)
print('Model trained:', type(model).__name__)
```

## 4. Evaluate the regression model

```
predictions = model.predict(X_test)
mae = mean_absolute_error(y_test, predictions)
mse = mean_squared_error(y_test, predictions)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, predictions)
print(f'MAE : {mae:.4f}')
print(f'MSE : {mse:.4f}')
print(f'RMSE: {rmse:.4f}')
print(f'R²  : {r2:.4f}')
```

## 5. Visualization

```
plt.figure(figsize=(8,5))
plt.scatter(y_test, predictions, alpha=0.7)
plt.xlabel('Actual delivery time (hours)')
plt.ylabel('Predicted delivery time (hours)')
plt.title('Actual vs Predicted Delivery Time')
lims=[min(y_test.min(), predictions.min()), max(y_test.max(), predictions.max())]
plt.plot(lims, lims, linestyle='--')
plt.tight_layout()
plt.show()
```

## 6. Export model for Flask

```
joblib.dump({'model': model, 'feature_columns': feature_columns, 'target_column': target_column, 'algorithm': 'Linear Regression'}, '../model/logistics_model.pkl')
print('Saved: ../model/logistics_model.pkl')
```
